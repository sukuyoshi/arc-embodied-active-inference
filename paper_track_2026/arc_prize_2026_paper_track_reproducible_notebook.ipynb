## 3. Verify Submission Integrity and Metrics Telemetry

```python
sub_candidates = [
    Path('submission.json'),
    Path('../embodied_core_engine/artifacts/submission.json'),
    Path('/kaggle/working/submission.json'),
]

for sc in sub_candidates:
    if sc.exists():
        with open(sc, 'r', encoding='utf-8') as f:
            sub_data = json.load(f)
        size_kb = round(sc.stat().st_size / 1024, 2)
        print('Found valid submission:', sc)
        print('Total Tasks:', len(sub_data))
        print('File Size:', size_kb, 'KB')
        sample_key = list(sub_data.keys())[0]
        print('Sample Key:', sample_key)
        print('Sample Attempts:', sub_data[sample_key])
        break
```
